```python
data = response.json()

headers = data[0]
first_capture = data[1]

capture = dict(zip(headers, first_capture))

print(capture)
```

```python
timestamp = capture["timestamp"]
original_url = capture["original"]

archived_url = (
    "https://web.archive.org/web/"
    + timestamp
    + "/"
    + original_url
)

print(archived_url)
```

```python
import requests

page_response = requests.get(archived_url)

print(page_response.status_code)
print(page_response.text[:500])
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(page_response.text, "html.parser")

page_text = soup.get_text(separator=" ", strip=True)

print(page_text[:1000])
```

```python
import re
from difflib import SequenceMatcher

# 1. Select the TypePad entry
entry = normal_posts[0]

# 2. Extract a field, including multiline BODY content
def get_field(entry, field):
    pattern = rf"^{re.escape(field)}:\s*(.*?)(?=^[A-Z][A-Z _-]*:|\Z)"
    match = re.search(pattern, entry, re.MULTILINE | re.DOTALL)

    if match:
        return match.group(1).strip()
    return ""

title = get_field(entry, "TITLE")
archive_body = get_field(entry, "BODY")

# 3. Display what we extracted
print("Title:", title)
print("Original BODY length:", len(archive_body))
print("Wayback text length:", len(page_text))
print("\nOriginal BODY preview:")
print(archive_body[:500])

# 4. Normalize text before comparing
def normalize(text):
    text = text.lower()
    text = re.sub(r"\s+", " ", text)
    return text.strip()

original = normalize(archive_body)
wayback = normalize(page_text)

# 5. Compare the texts
print("\n--- COMPARISON RESULTS ---")

if not original:
    print("Result: UNCERTAIN — original BODY is empty.")
elif not wayback:
    print("Result: UNCERTAIN — Wayback text is empty.")
else:
    similarity = SequenceMatcher(
        None, original, wayback
    ).ratio()

    print(f"Text similarity: {similarity:.1%}")

    if original == wayback:
        print("Result: SAME — exact text match.")
    elif original in wayback:
        print("Result: BODY FOUND within Wayback text.")
    elif similarity >= 0.85:
        print("Result: POSSIBLY SAME — review differences.")
    else:
        print("Result: DIFFERENT OR EXTRA TEXT — review manually.")
```

```python
from bs4 import BeautifulSoup
import re

# Convert the original BODY's HTML into readable text
original_soup = BeautifulSoup(archive_body, "html.parser")
original_text = original_soup.get_text(" ", strip=True)

# Normalize both texts
def normalize(text):
    text = text.lower()
    text = re.sub(r"\s+", " ", text)
    return text.strip()

original_clean = normalize(original_text)
wayback_clean = normalize(page_text)

# Check whether the original content appears in Wayback
print("Original article text length:", len(original_clean))
print("Wayback page text length:", len(wayback_clean))

if original_clean in wayback_clean:
    print("Result: BODY FOUND in Wayback text.")
elif original_clean[:150] in wayback_clean:
    print("Result: PARTIAL MATCH — beginning of BODY found.")
else:
    print("Result: NO MATCH FOUND — inspect the texts.")

print("\nOriginal text preview:")
print(original_clean[:500])

print("\nWayback text preview:")
print(wayback_clean[:500])
```

```python
# Check how much of the original text appears in Wayback
if original_clean in wayback_clean:
    print("Result: SAME — entire original BODY found in Wayback.")
else:
    # Check whether individual sentences or chunks match
    words = original_clean.split()
    chunk_size = 20

    matched_chunks = 0
    total_chunks = 0

    for i in range(0, len(words), chunk_size):
        chunk = " ".join(words[i:i + chunk_size])
        total_chunks += 1

        if chunk in wayback_clean:
            matched_chunks += 1

    match_percentage = matched_chunks / total_chunks * 100

    print(f"Matching text chunks: {matched_chunks}/{total_chunks}")
    print(f"Chunk match percentage: {match_percentage:.1f}%")

    if match_percentage >= 80:
        print("Result: LIKELY SAME — most text chunks match.")
    elif match_percentage >= 40:
        print("Result: PARTIAL MATCH — review differences.")
    else:
        print("Result: UNCERTAIN — substantial differences detected.")
```
